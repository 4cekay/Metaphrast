# **Configuration**

Setting up main language model and embedding system:

```python
# models in use
central_model = "microsoft/Phi-3-mini-4k-instruct"
embed_model = "bge-small-en-v1.5"

# subject/domain
subject = "High School Chemistry"
domain = "Cooking and Culinary"

target_concept = "chemical equilibrium"

top_k = 3
chunk_words = 150
max_new_tokens = 700
temperature = 0.7
```

Loading central language model:

```python
tokenizer = AutoTokenizer.from_pretrained(central_model, trust_remote_code=True)
model = AutoModelForCausalLM.from_pretrained(
    central_model,
    torch_dtype=torch.float16,
    trust_remote_code=True,
    device_map="auto"
)
```

```python
model.eval()
print("Device: ", model.device)
```

```python
messages = [
    {"role": "user", "content": "Who are you?"},
]
inputs = tokenizer.apply_chat_template(
	messages,
	add_generation_prompt=True,
	tokenize=True,
	return_dict=True,
	return_tensors="pt",
).to(model.device)

outputs = model.generate(**inputs, max_new_tokens=256)
print(tokenizer.decode(outputs[0][inputs["input_ids"].shape[-1]:]))
```
